```python
import itertools
def return_percentage(df,split,method , type, race, gender ):
    if split is not None:
        length=len(df[(df['split'] == split)])
        
    else:
        length=len(df)

    if (method is not None) & (split is not None) & (type is not None) & (race is not None) & (gender is not None):
        return (len(df[(df['method'] == method) & (df['split'] == split) & (df['type']==type) & (df['race']==race) & (df['gender']==gender)])/length)*100

def get_all_percentages(df):
    results = []

    # Get all unique values for each column
    splits = df['split'].unique()
    methods = df['method'].unique()
    types = df['type'].unique()
    races = df['race'].unique()
    genders = df['gender'].unique()

    # Loop through every combination
    for split, method, type_, race, gender in itertools.product(splits, methods, types, races, genders):
        perc = return_percentage(df, split, method, type_, race, gender)
        if perc is not None and perc > 0:
            results.append({
                'split': split,
                'method': method,
                'type': type_,
                'race': race,
                'gender': gender,
                'percentage': perc
            })

    return pd.DataFrame(results)
    
```

```python
import os
import pandas as pd
import subprocess
import numpy as np
import soundfile as sf
from concurrent.futures import ProcessPoolExecutor, as_completed
from tqdm import tqdm


# ============================================================
# CONFIG
# ============================================================
rootdir = "/home/manik/Downloads/FakeAVCeleb_v1.2"              # base dir containing videos
audio_root = "/home/manik/Downloads/alligned_audio"     # base dir to save aligned audios
df_path = "/home/manik/Downloads/FakeAvCelebPreprocessed/df_matching_frames.csv"              # your DataFrame CSV file
sample_rate = 16000                       # Hz
fps = 25                                 # video frame rate
num_workers = 8                           # parallel workers


# ============================================================
# 1️⃣ LOAD DATAFRAME AND BUILD FULL PATHS
# ============================================================
df = pd.read_csv(df_path)

# build full video paths
df["full_path"] = df.apply(
    lambda r: os.path.join(
        rootdir,
        str(r["type"]),
        str(r["race"]),
        str(r["gender"]),
        str(r["source"]),
        str(r["path"])
    ),
    axis=1
)

# build output audio root folder keeping structure
df["audio_folder"] = df.apply(
    lambda r: os.path.join(
        audio_root,
        str(r["type"]),
        str(r["race"]),
        str(r["gender"]),
        str(r["source"])
    ),
    axis=1
)

# ensure folders exist
for folder in df["audio_folder"].unique():
    os.makedirs(folder, exist_ok=True)


# ============================================================
# 2️⃣ HELPER FUNCTIONS
# ============================================================
def get_video_duration(video_path):
    """Get duration (in seconds) using ffprobe"""
    cmd = [
        "ffprobe", "-v", "error", "-show_entries", "format=duration",
        "-of", "default=noprint_wrappers=1:nokey=1", video_path
    ]
    return float(subprocess.check_output(cmd).decode().strip())


def extract_aligned_audio(video_path):
    """Return time-aligned audio matching video duration, zero-filled"""
    duration = get_video_duration(video_path)
    expected_samples = int(duration * sample_rate)

    # extract raw audio
    cmd = [
        "ffmpeg", "-i", video_path, "-f", "f32le", "-acodec", "pcm_f32le",
        "-ac", "1", "-ar", str(sample_rate), "-"
    ]
    try:
        raw = subprocess.check_output(cmd, stderr=subprocess.DEVNULL)
        audio_data = np.frombuffer(raw, np.float32)
    except subprocess.CalledProcessError:
        audio_data = np.zeros(0, np.float32)

    # pad or truncate to match video duration
    audio = np.zeros(expected_samples, dtype=np.float32)
    n = min(len(audio_data), expected_samples)
    audio[:n] = audio_data[:n]
    return audio, duration


def process_row(row):
    """Extract aligned audio and save frame-based segments"""
    video_path = row["full_path"]
    out_folder = row["audio_folder"]
    base = os.path.splitext(os.path.basename(row["path"]))[0]

    try:
        audio, duration = extract_aligned_audio(video_path)
    except Exception as e:
        return f"❌ {base}: {e}"

    samples_per_frame = int(sample_rate / fps)
    num_frames = int(duration * fps)

    for i in range(num_frames):
        start = i * samples_per_frame
        end = start + samples_per_frame
        seg = audio[start:end]

        if len(seg) < samples_per_frame:
            seg = np.pad(seg, (0, samples_per_frame - len(seg)))

        out_path = os.path.join(out_folder, f"{base}_frame{i:05d}.wav")
        sf.write(out_path, seg, sample_rate)

    return f"✅ {base} ({num_frames} segments)"


# ============================================================
# 3️⃣ PARALLEL PROCESSING WITH PROGRESS BAR
# ============================================================
if __name__ == "__main__":
    print(f"Processing {len(df)} videos using {num_workers} workers...\n")

    with ProcessPoolExecutor(max_workers=num_workers) as executor:
        futures = [executor.submit(process_row, row) for _, row in df.iterrows()]

        for f in tqdm(as_completed(futures), total=len(futures), desc="Extracting audio", ncols=90):
            result = f.result()
            tqdm.write(result)

    print("\n🎯 All videos processed successfully!")
```

```python
import torch
import torchaudio
import os
import re

def load_full_audio_from_chunks(folder_path, base_name):
    """
    Load multiple sequential audio chunks (e.g., video_frame00001.wav, video_frame00002.wav, ...)
    and concatenate them into one continuous waveform.
    
    Args:
        folder_path (str): Directory containing the audio files.
        base_name (str): Common prefix (e.g., "video_frame").
    
    Returns:
        waveform (Tensor): Combined audio waveform of shape [channels, total_samples].
        sample_rate (int): Sample rate (all chunks must have the same one).
    """

    # Find all matching chunk files
    files = [f for f in os.listdir(folder_path) if f.startswith(base_name) and f.endswith(".wav")]
    
    # Sort files numerically based on frame index
    files.sort(key=lambda x: int(re.findall(r'\d+', x)[-1]))

    waveforms = []
    sample_rate = None

    for f in files:
        path = os.path.join(folder_path, f)
        waveform, sr = torchaudio.load(path)

        if sample_rate is None:
            sample_rate = sr
        elif sample_rate != sr:
            raise ValueError(f"Sample rate mismatch in {f}: expected {sample_rate}, got {sr}")

        waveforms.append(waveform)

    # Concatenate along time dimension
    full_waveform = torch.cat(waveforms, dim=1)

    return full_waveform, sample_rate


import time
```
